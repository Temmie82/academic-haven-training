### 2. Create and Seed Data
We will create `books`, `customers`, and `orders`.

```python
%%sql
CREATE TABLE books (id INTEGER PRIMARY KEY, title TEXT, price REAL);
INSERT INTO books (title, price) VALUES ('1984', 8.50), ('The Hobbit', 15.00);

CREATE TABLE customers (id INTEGER PRIMARY KEY, name TEXT);
INSERT INTO customers (name) VALUES ('Alice'), ('Bob');

CREATE TABLE orders (customer_id INTEGER, book_id INTEGER);
INSERT INTO orders VALUES (1, 1), (2, 2);
```

### 3. The JOIN Query
Run this to see how we link the tables together.

```python
%%sql
SELECT customers.name, books.title
FROM orders
JOIN customers ON orders.customer_id = customers.id
JOIN books ON orders.book_id = books.id;
```
