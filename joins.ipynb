# SQL: Advanced Joins & NULL Handling

### 1. Setup
We will refresh our environment and add some 'lonely' data (data with no matches).

```python
%load_ext sql
%sql sqlite://

%%sql
CREATE TABLE users (u_id INTEGER PRIMARY KEY, username TEXT);
CREATE TABLE orders (order_id INTEGER PRIMARY KEY, user_id INTEGER, amount REAL);

INSERT INTO users VALUES (1, 'TechGuru'), (2, 'DataWizard'), (3, 'NewUser_NoOrders');
INSERT INTO orders (user_id, amount) VALUES (1, 500.00), (2, 150.00);
```

### 2. The LEFT JOIN Difference
If we used a regular JOIN, 'NewUser_NoOrders' would vanish. Watch how LEFT JOIN keeps them.

```python
%%sql
SELECT users.username, orders.amount
FROM users
LEFT JOIN orders ON users.u_id = orders.user_id;
```

### 3. Finding the Gaps
We can use `WHERE ... IS NULL` to specifically find customers who haven't spent any money yet.

```python
%%sql
SELECT username 
FROM users
LEFT JOIN orders ON users.u_id = orders.user_id
WHERE orders.amount IS NULL;
```

### 4. Cleaning up NULLs with COALESCE
Sometimes `NULL` looks messy in a report. We can use `COALESCE` to replace it with a default value like `0`.

```python
%%sql
SELECT username, COALESCE(amount, 0) AS spend_amount
FROM users
LEFT JOIN orders ON users.u_id = orders.user_id;
```
