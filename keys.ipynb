### 2. Creating Relational Tables
Note how `orders` uses `FOREIGN KEY` to reference the other tables.

```python
%%sql
CREATE TABLE products (
    p_id INTEGER PRIMARY KEY, 
    name TEXT, 
    price REAL
);

CREATE TABLE users (
    u_id INTEGER PRIMARY KEY, 
    username TEXT
);

CREATE TABLE sales (
    sale_id INTEGER PRIMARY KEY,
    user_id INTEGER,
    product_id INTEGER,
    FOREIGN KEY (user_id) REFERENCES users(u_id),
    FOREIGN KEY (product_id) REFERENCES products(p_id)
);

INSERT INTO products VALUES (1, 'Laptop', 1200), (2, 'Mouse', 25), (3, 'Keyboard', 75);
INSERT INTO users VALUES (1, 'TechGuru'), (2, 'DataWizard');
INSERT INTO sales (user_id, product_id) VALUES (1, 1), (1, 2), (2, 3);
```

### 3. The Power of the JOIN
Let's see who bought what and for how much.

```python
%%sql
SELECT 
    u.username, 
    p.name AS item_purchased, 
    p.price
FROM sales s
JOIN users u ON s.user_id = u.u_id
JOIN products p ON s.product_id = p.p_id;
```
